## 1. What is an Embedding?- An **embedding** is a list of numbers (a vector) that represents the *meaning* of a piece of text.- Similar sentences → similar (close) vectors.- Different sentences → far apart vectors.Example:- "I love dogs" and "I like puppies" → vectors will be **close**- "I love dogs" and "The stock market crashed" → vectors will be **far apart**We will use the `sentence-transformers` library, which uses free Hugging Face models.

### 7. Visualize Embeddings in 2D (using PCA)Embeddings have 384 dimensions — we can't see that. So we reduce it to 2D just to *visualize* how sentences cluster.

```python
# Exercise 1: Use 10 custom documents and test semantic search
custom_documents = [
    "Python is widely used for data science.",
    "Neural networks are important in deep learning.",
    "Football is played by two teams.",
    "The weather is rainy today.",
    "Machine learning finds patterns in data.",
    "Cats are popular household pets.",
    "The stock market changes every day.",
    "Artificial intelligence is used in many applications.",
    "Paris is a famous city in France.",
    "Basketball is played with a ball and a hoop."
]

custom_embeddings = model.encode(custom_documents)

def custom_semantic_search(query, top_k=3):
    query_embedding = model.encode([query])
    scores = cosine_similarity(query_embedding, custom_embeddings)[0]
    ranked_idx = scores.argsort()[::-1][:top_k]
    print(f"Query: {query}\n")
    for idx in ranked_idx:
        print(f"Score: {scores[idx]:.3f} | {custom_documents[idx]}")

custom_semantic_search("How are computers used to learn from data?")
```

```python
# Exercise 3: Build a small FAQ bot
faq_questions = [
    "What is Python?",
    "What is machine learning?",
    "What is an embedding?",
    "What is cosine similarity?",
    "What is semantic search?"
]

faq_answers = [
    "Python is a popular programming language used for many applications, including AI and data science.",
    "Machine learning is a method in which models learn patterns from data to make predictions or decisions.",
    "An embedding is a numeric vector representation that captures useful information about the meaning of data such as text.",
    "Cosine similarity measures how closely two vectors point in the same direction.",
    "Semantic search finds relevant information by meaning rather than only matching exact words."
]

faq_embeddings = model.encode(faq_questions)

def faq_bot(question):
    query_embedding = model.encode([question])
    scores = cosine_similarity(query_embedding, faq_embeddings)[0]
    best_idx = scores.argmax()
    print("Question:", question)
    print("Best match:", faq_questions[best_idx])
    print("Similarity:", round(float(scores[best_idx]), 3))
    print("Answer:", faq_answers[best_idx])

faq_bot("How does a computer learn patterns from examples?")
```
