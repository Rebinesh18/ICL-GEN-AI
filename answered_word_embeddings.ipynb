Word Embedding using One-Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
words = [['cat'],['dog'],['apple'],['car']]
```

```python
encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)
```

```python
print("words :",encoder.categories_[0])
print("Embeddings:")
print(result)
```

```python
from gensim.models import Word2Vec
```

```python
sentences = [["python","is","a","language"],
 ["java","is","a","language"],
  ["python","is","easy"],
   ["java","is","easy"],
    ["python","is","powerful"],
     ["java","is","powerful"]]
```

```python
model = Word2Vec(sentences,vector_size=10,window=2,min_count=1,workers=1)
```

```python
print(model.wv.most_similar('java',topn=3))
```

```python
print(model.wv.most_similar('python',topn=4))
```

## Completed Answers / Notes

- One-hot encoding represents each word as a binary vector with one `1` and the remaining values `0`.
- For the four words, the encoded order follows the encoder's sorted categories: `apple`, `car`, `cat`, `dog`.
- Word2Vec learns dense vectors from the example sentences. `most_similar()` returns the words whose learned vectors are closest to the requested word.
- Because this is a very small training corpus, the exact similarity ranking can vary with the installed Gensim version and random initialization.
